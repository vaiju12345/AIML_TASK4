```python
import numpy as np
from scipy import stats


marks = np.array([70, 80, 90, 80, 60, 85, 95, 80, 75, 85])

# Mean
print(f"Mean: {np.mean(marks):.2f}")

# Median
print(f"Median: {np.median(marks):.2f}")

# Mode
print(f"Mode: {stats.mode(marks, keepdims=False).mode}")

# Variance
print(f"Variance: {np.var(marks):.2f}")

# Standard Deviation
print(f"Standard Deviation: {np.std(marks):.2f}")
```
